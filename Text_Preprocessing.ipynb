## 1. Import Required Libraries

```python
import re
import nltk
import pandas as pd
import numpy as np

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

## 2. Sample Text Data

```python
texts = [
    "I love this product!!! It's amazing 😊",
    "This is the WORST purchase I've ever made... Never buying again!!!",
    "The food was okay, but the service was terrible. Visit: https://example.com",
    "Wow... what a great movie! Highly recommended. #BestMovie",
    "I'm not happy with the delivery. Contact me at test@email.com"
]

df = pd.DataFrame({'text': texts})
df.head()
```

## 3. Lowercasing

```python
df['text_lower'] = df['text'].str.lower()
df[['text', 'text_lower']]
```

## 4. Remove URLs

```python
def remove_urls(text):
    return re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)

df['text_no_url'] = df['text_lower'].apply(remove_urls)
df[['text_lower', 'text_no_url']]
```

## 5. Remove Email Addresses

```python
def remove_emails(text):
    return re.sub(r'\S+@\S+', '', text)

df['text_no_email'] = df['text_no_url'].apply(remove_emails)
df[['text_no_url', 'text_no_email']]
```

## 6. Remove Punctuation & Special Characters

```python
def remove_punctuation(text):
    return re.sub(r'[^\w\s]', '', text)

df['text_clean'] = df['text_no_email'].apply(remove_punctuation)
df[['text_no_email', 'text_clean']]
```

## 7. Remove Numbers

```python
def remove_numbers(text):
    return re.sub(r'\d+', '', text)

df['text_no_num'] = df['text_clean'].apply(remove_numbers)
df[['text_clean', 'text_no_num']]
```

## 8. Tokenization

```python
df['tokens'] = df['text_no_num'].apply(word_tokenize)
df[['text_no_num', 'tokens']]
```

## 9. Remove Stopwords

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

df['tokens_no_stop'] = df['tokens'].apply(remove_stopwords)
df[['tokens', 'tokens_no_stop']]
```

## 10. Stemming

```python
stemmer = PorterStemmer()

def stemming(tokens):
    return [stemmer.stem(word) for word in tokens]

df['stemmed'] = df['tokens_no_stop'].apply(stemming)
df[['tokens_no_stop', 'stemmed']]
```

## 11. Lemmatization (Better than Stemming)

```python
lemmatizer = WordNetLemmatizer()

def lemmatize(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

df['lemmatized'] = df['tokens_no_stop'].apply(lemmatize)
df[['tokens_no_stop', 'lemmatized']]
```

## 12. Final Clean Text

```python
df['final_text'] = df['lemmatized'].apply(lambda x: ' '.join(x))
df[['text', 'final_text']]
```

## 13. Complete Preprocessing Function

```python
def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()
    
    # 2. Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    
    # 3. Remove Emails
    text = re.sub(r'\S+@\S+', '', text)
    
    # 4. Remove Punctuation
    text = re.sub(r'[^\w\s]', '', text)
    
    # 5. Remove Numbers
    text = re.sub(r'\d+', '', text)
    
    # 6. Tokenize
    tokens = word_tokenize(text)
    
    # 7. Remove Stopwords
    tokens = [word for word in tokens if word not in stop_words]
    
    # 8. Lemmatize
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    # 9. Join back to string
    return ' '.join(tokens)

# Apply on original text
df['cleaned'] = df['text'].apply(preprocess_text)
df[['text', 'cleaned']]
```

## 14. Test with New Sentence

```python
sample = "Wow!!! This product is SUPER amazing... Visit https://shop.com or email me@test.com 😂"
print("Original :", sample)
print("Cleaned  :", preprocess_text(sample))
```
